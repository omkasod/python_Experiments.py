Case Study 1: Write a function to calculate average marks from a list.

```python
def calculate_average(marks):
  if len(marks)==0:
    return 0

  total=sum(marks)
  average=total/len(marks)
  return average

# Example usage
marks_list=[55,95,67,88,99]
avg=calculate_average(marks_list)

print("Average marks:",avg)
```

Case Study 2: Write a function using positional arguments to print name and salary.

```python
def employee_details(name,salary):
  print("Name:",name)
  print("Salary:",salary)

# Function call using positional arguments
employee_details("Rahul",50000)
```

Case Study 3: Write a function using args to accept multiple prices

```python
def total_prices(*prices):
  print("Prices:",prices)
  print("Total Price:",sum(prices))

# Example function call
total_prices(100,250,399,150)
```

Case Study 4: Use **kwargs to accept different student attributes.

```python
def student_details(**kwargs):
  for key,value in kwargs.items():
    print(f"{key}: {value}")

# Example function call
student_details(name="Mahi",age=44,course="Python",marks=85)
```

Case Study 5: Use a lambda function to calculate discounted price.

```python
# Lambda function to calculate discounted price
discounted_price = lambda price,discount:price-(price*discount/100)

# Example usage
final_price=discounted_price(1000,20)

print("Discounted Price:",final_price)
```

Case Study 6: Use a nested function to check password validity.

```python
def password_checker(password):
  def is_valid():

    # Check password conditions
    return len(password) >= 8 and any(char.isdigit()for char in password)
  if is_valid():
    print("Password is valid")

  else:
    print("Password is invalid")


# Example usage
password_checker("Pass1234")
```

Case Study 7: Write a function returning multiple values.

```python
def calculate(a,b):
  addition =  a + b
  subtraction = a - b
  multiplication = a * b

  return addition, subtraction, multiplication

# Example usage
add,sub,mul=calculate(10,5)
print("Addition:", add)
print("Subtraction:", sub)
print("Multiplication:", mul)
```
