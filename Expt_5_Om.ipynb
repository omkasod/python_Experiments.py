Q1. A teacher want to store marks of students in list

- Add new marks
- Find highest & lowest marks
- Sort the marks
- Count students scoring above 50

```python
# A teacher wants to store marks of students in a list
marks=[45,67,89,34,76]

# Add new marks
marks.append(55)

# Highest & Lowest
print("Max:",max(marks))
print("Min:",min(marks))

# Sorting
marks.sort()
print("Sorted:",marks)

# Count above 50
count=0
for m in marks:
  if m > 50:
    count += 1
print("Students above 50:",count)
```

Q3. Store salaries in a list:

- Increase all salaries by 10%
- Find average salary
- Display updated list

```python
salaries=[20000,30000,25000,40000]

# Increase salary by 10%
for i in range(len(salaries)):
  salaries[i] = salaries[i] * 1.10

# Average salary
avg = sum(salaries)/len(salaries)

print("Updated salaries:",salaries)
print("Average salary:",avg)
```

Q4. Given a list of numbers:

- Separate even and odd numbers
- Find sum of even numbers
- Reverse the list

```python
nums=[10,15,20,25,30]

even=[]
odd=[]

for n in nums:
  if n % 2 == 0:
    even.append(n)
  else:
    odd.append(n)

print("Even:",even)
print("Odd:",odd)

print("Sum of even:",sum(even))
nums.reverse()
print("Reversed:",nums)
```

Q5. A college wants to manage student data by using list:

- Store student marks
- Add new student marks
- Remove failed students (<40)
- Sort marks
- Display highest, lowest and average
- Count distinction students(>75)

```python
marks=[55,78,34,89,62,41]

# Add new marks
marks.append(73)

# Remove failed students
marks=[m for m in marks if m >= 40]

# Sort marks
marks.sort()

# Highest, lowest, average
print("Highest:", max(marks))
print("Lowest:", min(marks))
print("Average:", sum(marks)/len(marks))

# Count distinction students
count = 0
for m in marks:
  if m > 75:
    count += 1
print("Distinction students:",count)
print("Final list:",marks)
```
