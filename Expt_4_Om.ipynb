Case Study 1: FOR Loop (5 Programs)

```python
# Program 1: Print Numbers from 1 to 10 Using for Loop
print("Program to Print Numbers from 1 to 10")
print("--------------------------------------")

for number in range(1,11):
  print("Current Number is:",number)

print("--------------------------------------")
print("All numbers printed successfully")
```

```python
# Program 2: Sum of First 10 Numbers Using for Loop

print("Program to Find Sum of First 10 Numbers")
print("--------------------------------------")

sum_of_numbers = 0

for number in range(1,11):
  print("Adding:",number)
  sum_of_numbers=sum_of_numbers+number
  print("Current Sum:",sum_of_numbers)

print("--------------------------------------")
print("Final Sum of First 10 Numbers is:",sum_of_numbers)
```

```python
# Program 3: Multiplication Table Using for Loop

print("Program to Print Multiplication Table")
print("--------------------------------------")

number = int(input("Enter a Number: "))

print("--------------------------------------")
print("Multiplication Table of", number)
print("--------------------------------------")

for i in range(1, 11):
  result = number * i
  print(number, "x", i, "=", result)

print("--------------------------------------")
print("Table Printed Successfully")
```

```python
# Program 4: Factorial of a Number Using for Loop

print("Program to Find Factorial")
print("--------------------------------------")

number=int(input("Enter a Number: "))

factorial = 1

if number < 0:
  print("Factorial does not exist for negative numbers")

elif number==0:
  print("Factorial of 0 is 1")

else:
  for i in range(1,number + 1):
    factorial = factorial * i
    print("Current Value After Multiplying",i,"is",factorial)

  print("--------------------------------------")
  print("Final Factorial of",number,"is:",factorial)
```

```python
# Program 5: Print Even Numbers Using for Loop

print("Program to Print Even Numbers")
print("--------------------------------------")

starting_number = 1
ending_number = 20

print("Even Numbers Between",starting_number,"and",ending_number)
print("--------------------------------------")

for number in range(starting_number,ending_number+1):

  if number % 2 == 0:
    print(number,"is an Even Number")

print("--------------------------------------")
print("All Even Numbers Printed Successfully")
```

```python
# Program 6: Print Numbers from 1 to 10 Using while Loop

print("Program to Print Numbers from 1 to 10")
print("--------------------------------------")

number = 1

while number <= 10:
  print("Current Number is:", number)
  number=number + 1

print("--------------------------------------")
print("All Numbers Printed Successfully")
```

```python
# Program 7: Reverse a Number Using while Loop

print("Program to Reverse a Number")
print("--------------------------------------")

number=int(input("Enter a Number: "))

original_number=number
reverse_number=0

while number > 0:
  digit = number % 10
  reverse_number=reverse_number*10+digit
  number = number//10

  print("Current Reversed Number:",reverse_number)

print("--------------------------------------")
print("Original Number:",original_number)
print("Reversed Number:",reverse_number)
```

```python
# Program 8: Count Digits in a Number Using while Loop

print("Program to Count Digits")
print("--------------------------------------")

number = int(input("Enter a Number: "))

count = 0
temporary_number = number

while temporary_number > 0:

  temporary_number = temporary_number//10
  count = count + 1

print("Current Digit Count:",count)
print("--------------------------------------")
print("Total Number of Digits in",number,"is:",count)
```

```python
# Program 9: Check Palindrome Number Using while Loop

print("Program to Check Palindrome Number")
print("--------------------------------------")

number=int(input("Enter a Number: "))

original_number=number
reverse_number=0

while number > 0:

  digit=number%10
  reverse_number=reverse_number*10+digit
  number=number//10

print("--------------------------------------")
print("Original Number:",original_number)
print("Reversed Number:",reverse_number)

if original_number==reverse_number:
  print("The Number is a Palindrome")

else:
  print("The Number is Not a Palindrome")
```

```python
# Program 10: Print Odd Numbers Using while Loop

print("Program to Print Odd Numbers")
print("--------------------------------------")

number = 1

while number <= 20:
  if number % 2 != 0:
    print(number,"is an Odd Number")

  number = number + 1

print("--------------------------------------")
print("All Odd Numbers Printed Successfully")
```

Case Study 3 : Nested FOR Loop (5 Programs)

```python
# Program 1: Print Square Pattern Using Nested Loop

print("Program to Print Square Pattern")
print("--------------------------------")

rows = 5
columns = 5

for i in range(rows):
  for j in range(columns):
    print("*",end=" ")

  print()

print("--------------------------------")
print("Square Pattern Printed Successfully")
```

```python
# Program 2: Print Right Triangle Pattern Using Nested Loop

print("Program to Print Right Triangle Pattern")
print("----------------------------------------")

number_of_rows=5

for i in range(1,number_of_rows+1):
  for j in range(i):
    print("*",end=" ")

  print()

print("----------------------------------------")
print("Triangle Pattern Printed Successfully")
```

```python
# Program 3: Multiplication Tables from 1 to 5 Using Nested Loop

print("Program to Print Multiplication Tables")
print("--------------------------------------")

for i in range(1, 6):
  print("\nMultiplication Table of", i)
  print("----------------------------")

for j in range(1, 11):

  result = i * j
  print(i, "x", j, "=", result)

print("--------------------------------------")
print("All Tables Printed Successfully")
```

```python
# Program 4: Print Number Pattern Using Nested Loop

print("Program to Print Number Pattern")
print("--------------------------------")

rows = 5

for i in range(1,rows+1):
  for j in range(1,i+1):
    print(j,end=" ")

  print()

print("--------------------------------")
print("Number Pattern Printed Successfully")
```

```python
# Program 5: Print Pyramid Star Pattern Using Nested Loop

print("Program to Print Pyramid Pattern")
print("--------------------------------")

rows = 5

for i in range(1,rows+1):

  # Print Spaces
  for j in range(rows-i):
    print(" ",end=" ")

  # Print Stars
  for k in range(2*i-1):
    print("*",end=" ")

  print()
print("--------------------------------")
print("Pyramid Pattern Printed Successfully")
```

Case Study 4 : while-else Loop (5 Programs)

```python
# Program 1: Print Numbers Using while-else

print("Program to Print Numbers from 1 to 5")
print("------------------------------------")

number = 1

while number <= 5:
    print("Current Number is:", number)
    number = number + 1

else:
    print("------------------------------------")
    print("Loop Finished Successfully")

print("------------------------------------")
print("Program Ended")
```

```python
# Program 2: Sum of Numbers Using while-else

print("Program to Find Sum of Numbers")
print("--------------------------------")

number = 1
sum_of_numbers = 0

while number <= 10:

  sum_of_numbers=sum_of_numbers + number

  print("Adding:",number)
  print("Current Sum:",sum_of_numbers)

  number = number + 1

else:
  print("--------------------------------")
  print("Loop Completed Successfully")
print("Final Sum is:",sum_of_numbers)
```

```python
# Program 3: Check Even and Odd Numbers Using while-else

print("Program to Check Even and Odd Numbers")
print("--------------------------------------")

number = 1

while number <= 10:

  if number%2==0:
    print(number,"is Even")

  else:
    print(number,"is Odd")

  number=number+1

else:
  print("--------------------------------------")
  print("All Numbers Checked Successfully")
```

```python
# Program 4: Password Checking Using while-else

print("Program for Password Verification")
print("----------------------------------")

correct_password = "python123"

attempts = 1

while attempts <= 3:

  entered_password=input("Enter Password: ")

  if entered_password == correct_password:
    print("Correct Password")
    break

  else:
    print("Wrong Password")
    attempts=attempts+1
else:
  print("----------------------------------")
  print("Maximum Attempts Reached")
  print("Access Denied")

print("----------------------------------")
print("Program Finished")
```

```python
# Program 5: Search Number in List Using while-else

print("Program to Search Number in List")
print("--------------------------------")

numbers = [10,20,30,40,50]

search_number = int(input("Enter Number to Search: "))

index = 0

while index < len(numbers):

  if numbers[index] == search_number:
    print(search_number,"Found at Position",index)
    break

  index = index + 1

else:
  print("--------------------------------")
  print(search_number,"Not Found in the List")

print("--------------------------------")
print("Program Completed")
```

Case Study 5: Loop Control continue statement( 5 Programs)

```python
# Program 1: Skip Even Numbers Using continue Statement

print("Program to Skip Even Numbers")
print("--------------------------------")

for number in range(1,11):

  if number % 2 == 0:
    continue

  print(number,"is an Odd Number")

print("--------------------------------")
print("Program Completed Successfully")
```

```python
# Program 2: Skip Number 5 Using continue Statement

print("Program to Skip Number 5")
print("--------------------------------")

for number in range(1,11):
  if number==5:
    continue

  print("Current Number is:", number)

print("--------------------------------")
print("Number 5 was Skipped Successfully")
```

```python
# Program 3: Print Characters Except Spaces Using continue Statement

print("Program to Ignore Spaces in String")
print("-----------------------------------")

text = "Python Programming Language"

for character in text:

  if character == " ":
    continue

  print(character)

print("-----------------------------------")
print("All Characters Printed Successfully")
```

```python
# Program 4: Skip Negative Numbers Using continue Statement

print("Program to Skip Negative Numbers")
print("---------------------------------")

numbers = [10,-5, 20,-15, 30,-25, 40]

for number in numbers:

  if number < 0:
    continue

  print(number, "is a Positive Number")

print("---------------------------------")
print("Negative Numbers Skipped Successfully")
```

```python
# Program 5: Skip Multiples of 3 Using continue Statement

print("Program to Skip Multiples of 3")
print("--------------------------------")

for number in range(1,21):

  if number % 3 == 0:
    continue

  print(number)

print("--------------------------------")
print("Multiples of 3 Skipped Successfully")
```

```python
# Program 1: Stop Loop at Number 5 Using break Statement

print("Program to Stop Loop at Number 5")
print("--------------------------------")

for number in range(1,11):

  if number==5:
    print("Number 5 Found")
    print("Loop is Terminating")
    break

  print("Current Number is:",number)

print("--------------------------------")
print("Program Completed Successfully")
```

```python
# Program 2: Password Verification Using break Statement

print("Program for Password Verification")
print("----------------------------------")

correct_password="python123"

while True:

  entered_password = input("Enter Password: ")

  if entered_password==correct_password:
    print("Correct Password Entered")
    print("Access Granted")
    break

else:
  print("Wrong Password")
  print("Try Again")

print("----------------------------------")
print("Program Finished")
```

```python
# Program 3: Search Element in List Using break Statement

print("Program to Search Element in List")
print("----------------------------------")

numbers=[10,20,30,40,50]

search_number=int(input("Enter Number to Search: "))

for number in numbers:

  if number == search_number:
    print(search_number,"Found in the List")
    break
  print(number,"Checked")

print("----------------------------------")
print("Program Completed")
```

```python
# Program 4: Stop at First Negative Number Using break Statement

print("Program to Stop at First Negative Number")
print("-----------------------------------------")

numbers=[10,20,30,-5,40,50]

for number in numbers:

  if number<0:
    print("Negative Number Found:",number)
    print("Loop Terminated")
    break

  print("Positive Number:",number)

print("-----------------------------------------")
print("Program Completed Successfully")
```

```python
# Program 5: Multiplication Table Until User Stops Using break Statement

print("Program to Print Multiplication Tables")
print("-------------------------------------")

while True:

    number = int(input("Enter a Number: "))

    print("-------------------------------------")
    print("Multiplication Table of", number)

    for i in range(1, 11):
        print(number, "x", i, "=", number * i)

    print("-------------------------------------")

    choice = input("Do You Want to Continue? (yes/no): ")

    if choice.lower() == "no":
        print("Loop Terminated by User")
        break

print("-------------------------------------")
print("Program Finished Successfully")
```

Case Study 7: Loop Control pass statement( 5 Programs)

```python
# Program 1: Using pass in Empty Loop Body

print("Program to Demonstrate pass in Loop")
print("-----------------------------------")

for number in range(1,6):
  if number==3:
    pass # Nothing happens when number is 3

  print("Current Number is:",number)

print("-----------------------------------")
print("Program Completed Successfully")
```

```python
# Program 2: Placeholder Function Using pass

print("Program to Demonstrate pass in Function")
print("---------------------------------------")

def future_function():
  pass      # Function not implemented yet

print("Function Created but Not Implemented")

future_function()

print("---------------------------------------")
print("Program Completed Successfully")
```

```python
# Program 3: Using pass in Conditional Statement

print("Program Using pass in Condition")
print("--------------------------------")

number = 10

if number > 0:
  pass        # Future logic can be added here

print("Number is:",number)

print("--------------------------------")
print("Program Completed Successfully")
```

```python
 # Program 4: Empty Class Using pass

print("Program to Demonstrate Empty Class")

class Student:
  pass      # Class structure defined but no content yet

s1 = Student()

print("Student Object Created Successfully")

print("--------------------------------")
print("Program Completed Successfully")
```

```python
# Program 5: Menu Placeholder Using pass

print("Program for Menu System")
print("--------------------------------")

choice = 1

if choice == 1:
  print("Option 1 Selected")

elif choice == 2:
  pass        # Future option will be added here

elif choice == 3:
  pass # Future option will be added here

else:
  print("Invalid Choice")

print("--------------------------------")
print("Program Completed Successfully")
```
