Case Study 1

```python
# File Handling Example

# Step 1: Writing to file
file=open("student.txt","w")
file.write("Name: Mahendra Singh Dhoni\n")
file.write("Department: Robotics and Automation\n")
file.write("College: Zeal Engineering College\n")
file.close()
print("successfully.")

# Step 2: Reading from file
file=open("student.txt","r")
content=file.read()
print("\nFile Content:")
print(content)
file.close()
```

Case Study 2: Appending Data to Existing File

```python
file=open("student.txt","a")
file.write("Course: Python Programming\n")
file.close()
print("Data appended successfully.")
```

Case Study 3: Reading Line by Line

```python
file=open("student.txt","r")
for line in file:
  print(line)
file.close()
```

Case Study 4: Using with open()

```python
with open("sample.txt", "w") as file:
  file.write("Hello Python")
with open("sample.txt", "r") as file:
  data = file.read()
  print(data)
```

```python
with open("attendance.txt", "w") as file:
  file.write("Vijay - Present\n")
  file.write("Amit - Absent\n")
  file.write("Neha - Present\n")
with open("attendance.txt", "r") as file:
  print(file.read())
```
